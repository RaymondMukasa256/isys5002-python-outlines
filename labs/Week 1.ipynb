### Comments

A line beginning with `#` is a comment. Python ignores comments, so they are useful for explaining code.

```python
# This line is a comment and will not be executed.
print("Hello, World!")
print()
# \n starts a new line inside a string.
print("This is the first line. \nThis is the second line.")
# Triple quotes preserve line breaks.
print("""Three quotes can also be used
to print multiple lines using a single print function""")
```
