```python
def average(readings):
    """Calculates the mean of a list of readings."""
    if not readings:
        return 0
    return sum(readings) / len(readings)

def count_above(readings, threshold):
    """Counts how many readings are above a given threshold."""
    count = 0
    for reading in readings:
        if reading > threshold:
            count += 1
    return count

def print_daily_advice(temperatures_list):
    """Prints daily weather advice based on temperature."""
    print("\nDaily Advice:")
    for day_index, temp in enumerate(temperatures_list):
        day = day_index + 1
        if temp > 25:
            print(f"Day {day}: It's a hot day, stay hydrated!")
        elif temp >= 20:
            print(f"Day {day}: Pleasant weather, enjoy outside!")
        elif temp >= 15:
            print(f"Day {day}: A bit cool, a light jacket might be good.")
        else:
            print(f"Day {day}: Cold day, bundle up!")

# Main program
# 'temperatures' list is assumed to be available from previous cells

highest_temp = max(temperatures)
lowest_temp = min(temperatures)

avg_temp_raw = average(temperatures)
average_temp = round(avg_temp_raw, 1)

days_above_average = count_above(temperatures, average_temp)

print(f"Weekly Weather Report:")
print(f"Highest reading: {highest_temp}°C")
print(f"Lowest reading: {lowest_temp}°C")
print(f"Average reading: {average_temp}°C")
print(f"Number of days above average: {days_above_average}")

print_daily_advice(temperatures)
```

##Challenge: Weather Station

```python
# Install pyinputplus if not already installed
!pip install pyinputplus

import pyinputplus as pypi

def display_menu():
    """Displays the weather station menu options."""
    print("\n--- Weather Station Menu ---")
    print("1. Add a new reading")
    print("2. Display all readings")
    print("3. Show summary statistics")
    print("4. Get advice for a specific reading")
    print("5. Exit")
    print("----------------------------")

def add_reading(readings_list):
    """Prompts the user for a new temperature reading and adds it to the list."""
    try:
        # Using pyinputplus to ensure integer input within a reasonable range
        new_reading = pypi.inputInt("Enter temperature reading in Celsius (e.g., 25): ", min=-50, max=50)
        readings_list.append(new_reading)
        print(f"Reading {new_reading}°C added successfully.")
    except Exception as e:
        print(f"An error occurred during input: {e}. Please try again.")

def show_readings(readings_list):
    """Displays all current readings, numbered from 1."""
    if not readings_list:
        print("No readings to display yet. Add some readings first.")
        return

    print("\n--- All Readings ---")
    for i, reading in enumerate(readings_list):
        print(f"{i + 1}. {reading}°C")
    print("--------------------")

def calculate_average(readings_list):
    """Calculates the average of a list of readings."""
    if not readings_list:
        return 0
    return sum(readings_list) / len(readings_list)

def get_advice_for_reading(temp):
    """Returns weather advice based on a single temperature reading."""
    if temp > 25:
        return "It's a hot day, stay hydrated!"
    elif temp >= 20:
        return "Pleasant weather, enjoy outside!"
    elif temp >= 15:
        return "A bit cool, a light jacket might be good."
    else:
        return "Cold day, bundle up!"

def show_statistics(readings_list):
    """Displays summary statistics (count, average, highest, lowest) of readings."""
    if not readings_list:
        print("No readings available for statistics yet. Add some readings first.")
        return

    count = len(readings_list)
    average_val = calculate_average(readings_list)
    highest = max(readings_list)
    lowest = min(readings_list)

    print("\n--- Summary Statistics ---")
    print(f"Total readings: {count}")
    print(f"Average reading: {average_val:.1f}°C") # Rounded to one decimal place
    print(f"Highest reading: {highest}°C")
    print(f"Lowest reading: {lowest}°C")
    print("--------------------------")

def get_specific_advice_for_day(readings_list):
    """Asks the user for a specific reading number and provides advice for it."""
    if not readings_list:
        print("No readings available to give advice for yet.")
        return

    try:
        reading_num = pypi.inputInt(f"Enter the reading number (1 to {len(readings_list)}) for advice: ",
                                    min=1, max=len(readings_list))
        selected_reading = readings_list[reading_num - 1]
        advice = get_advice_for_reading(selected_reading)
        print(f"Advice for Reading {reading_num} ({selected_reading}°C): {advice}")
    except Exception as e:
        print(f"Invalid input: {e}. Please enter a valid reading number.")

def main():
    """Main function to run the weather station program loop."""
    all_readings = [] # Initialize an empty list to store readings
    while True:
        display_menu()
        try:
            choice = pypi.inputInt("Enter your choice (1-5): ", min=1, max=5)
        except Exception as e:
            print(f"Invalid input: {e}. Please enter a number between 1 and 5.")
            continue

        if choice == 1:
            add_reading(all_readings)
        elif choice == 2:
            show_readings(all_readings)
        elif choice == 3:
            show_statistics(all_readings)
        elif choice == 4:
            get_specific_advice_for_day(all_readings)
        elif choice == 5:
            print("Exiting Weather Station. Goodbye!")
            break

# Run the main program
if __name__ == "__main__":
    main()
```
