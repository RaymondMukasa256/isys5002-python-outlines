````python

#dictionary containing sales data
sales_data = {
    "P001": {"name": "Widget A", "price": 9.99, "qty_sold": 120},
    "P002": {"name": "Widget B", "price": 14.50, "qty_sold": 85},
    "P003": {"name": "Widget C", "price": 7.25, "qty_sold": 200},
    "P004": {"name": "Widget D", "price": 12.75, "qty_sold": 75}
}
````

````python
total_revenue = 0

for product_id, product_info in sales_data.items():
    price = product_info['price']
    qty_sold = product_info['qty_sold']
    product_revenue = price * qty_sold
    total_revenue += product_revenue

print(f"Total Revenue: ${total_revenue:.2f}")
````

## Calculating Total Revenue using Dictionary Operations

To programmatically calculate the total revenue from the given `sales_data` dictionary:

1.  **Initialize a `total_revenue` variable**: Start by creating a variable, for example, `totRevenue`, and set its initial value to `0`. This variable will accumulate the revenue from each product.

2.  **Iterate through the dictionary**: Use a loop (e.g., a `for` loop) to go through each `product_id` and its corresponding `product_info` (which is itself a dictionary) in the `sales_data` dictionary. The `.items()` method is useful here as it returns both the keys and values.

    ```python
    for product_id, product_info in sales_data.items():
        # ... calculations for each product
    ```

3.  **Extract `price` and `qty_sold`**: Inside the loop, for each `product_info` dictionary, access the `price` and `qty_sold` values using their respective keys. For example, `product_info['price']` and `product_info['qty_sold']`.

4.  **Calculate `product_revenue`**: Multiply the `price` by the `qty_sold` for the current product to get the `product_revenue`.

    ```python
    price = product_info['price']
    qty_sold = product_info['qty_sold']
    product_revenue = price * qty_sold
    ```

5.  **Accumulate `total_revenue`**: Add the calculated `product_revenue` to the `total_revenue` variable.

    ```python
    total_revenue += product_revenue
    ```

6.  **Display the result**: After the loop has finished iterating through all products, the `total_revenue` variable will hold the sum of all individual product revenues. You can then print this value, often formatted to two decimal places for currency.

    ```python
    print(f"Total Revenue: ${total_revenue:.2f}")
    ```

````python
highest_qty_sold = -1
product_with_highest_qty = None

for product_id, product_info in sales_data.items():
    qty_sold = product_info['qty_sold']
    if qty_sold > highest_qty_sold:
        highest_qty_sold = qty_sold
        product_with_highest_qty = product_info['name']

print(f"The product with the highest quantity sold is: {product_with_highest_qty} ({highest_qty_sold} units)")
````
